```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

auto = pd.read_csv('Auto.csv')
```

```python
display(auto.head(5))
display(auto['horsepower'].unique())
```

```python

auto = auto[auto['horsepower']!='?'].copy()
auto['horsepower'] = auto['horsepower'].astype(int)

display(auto.info())
display(auto.describe())
print(auto.shape)
```

```python
plt.figure()
plt.grid(alpha=0.3)
plt.scatter(auto['horsepower'],auto['mpg'],color='red',edgecolors='black')
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.show()
```
