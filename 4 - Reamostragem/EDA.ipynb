**Os códigos a seguir são referentes ao tratamento da base de dados e da plotagem da coluna "horsepower" em função da coluna "mpg". Embora não tenha sido feito pelos autores do livro, optei por realizar essa etapa para maior entendimento do problema.**

```python
import pandas as pd
import matplotlib.pyplot as plt

auto = pd.read_csv('Auto.csv')
```

```python
display(auto.head(5))
display(auto.info())
display(auto['horsepower'].unique())
```

```python

auto = auto[auto['horsepower']!='?'].copy()
auto['horsepower'] = auto['horsepower'].astype(int)

display(auto.info())
display(auto.describe())
print(auto.shape)
```

```python
plt.figure()
plt.grid(alpha=0.3)
plt.scatter(auto['horsepower'],auto['mpg'],color='red',edgecolors='black')
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.title('Ordem linear')
plt.show()
```

```python
plt.figure()
plt.grid(alpha=0.3)
plt.scatter(auto['horsepower']**(1/2),auto['mpg'],color='red',edgecolors='black')
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.title('Ordem quadrática')
plt.show()
```

```python
plt.figure()
plt.grid(alpha=0.3)
plt.scatter(auto['horsepower']**(1/3),auto['mpg'],color='red',edgecolors='black')
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.title('Ordem cúbica')
plt.show()
```
